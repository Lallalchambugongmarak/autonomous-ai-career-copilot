```python
class LLMService:
    def __init__(self, client, model: str):
        self.client = client
        self.model = model

    def generate(self, prompt: str) -> str:
        response = self.client.models.generate_content(
            model=self.model,
            contents=prompt,
        )
        return response.text


llm_service = LLMService(
    client=client,
    model=config.llm_model,
)

print("✓ LLM Service created")
```

```python
from pydantic import BaseModel, Field


class AgentResponse(BaseModel):
    task: str = Field(description="The task given to the agent")
    summary: str = Field(description="The agent's concise response")
    confidence: float = Field(
        ge=0.0,
        le=1.0,
        description="Confidence score between 0 and 1"
    )


print("✓ AgentResponse model created")
```

```python
response = client.models.generate_content(
    model=config.llm_model,
    contents="""
    You are an AI Research Agent.

    Task:
    Explain what Retrieval-Augmented Generation (RAG) is.

    Return a structured response containing:
    - task
    - summary
    - confidence

    The confidence value must be between 0 and 1.
    """,
    config={
        "response_mime_type": "application/json",
        "response_json_schema": AgentResponse.model_json_schema(),
    },
)

print(response.text)
```

```python
import json

agent_response = AgentResponse.model_validate(
    json.loads(response.text)
)

print("✓ Response validated successfully")
print()
print("Task:", agent_response.task)
print("Summary:", agent_response.summary)
print("Confidence:", agent_response.confidence)
```

```python
from google.genai import types
import json


def generate_structured_response(prompt: str, response_model):
    response = client.models.generate_content(
        model=config.llm_model,
        contents=prompt,
        config=types.GenerateContentConfig(
            response_mime_type="application/json",
            response_schema=response_model,
        ),
    )

    return response_model.model_validate(
        json.loads(response.text)
    )


print("✓ Structured LLM function created")
```

```python
from typing import Literal
from pydantic import BaseModel, Field


class AgentTask(BaseModel):
    task_id: str = Field(description="Unique identifier for the task")
    description: str = Field(description="What the agent needs to accomplish")
    agent: Literal[
        "research",
        "analysis",
        "retrieval",
        "verification",
        "synthesis",
        "automation"
    ] = Field(description="Agent responsible for the task")
    priority: Literal["low", "medium", "high"] = Field(
        default="medium",
        description="Task priority"
    )


print("✓ AgentTask model created")
```

```python
from pydantic import BaseModel, Field


class AgentPlan(BaseModel):
    objective: str = Field(
        description="The main objective of the user's request"
    )
    tasks: list[AgentTask] = Field(
        description="Ordered list of tasks required to complete the objective"
    )


print("✓ AgentPlan model created")
```

```python
user_request = """
Compare the latest AI agent frameworks.
Research the major frameworks, analyze their architecture,
verify important technical claims, and prepare a concise report.
"""

agent_plan = generate_structured_response(
    prompt=f"""
You are the Orchestrator Agent in an AI multi-agent system.

Your job is to break the user's request into clear tasks
and assign each task to the appropriate specialized agent.

Available agents:
- research
- analysis
- retrieval
- verification
- synthesis
- automation

User request:
{user_request}

Create an ordered execution plan.
Do not perform the tasks yourself.
""",
    response_model=AgentPlan,
)

print("✓ Agent plan generated")
print()
print("Objective:")
print(agent_plan.objective)
print()
print("Tasks:")

for task in agent_plan.tasks:
    print(
        f"- {task.task_id} | "
        f"{task.agent} | "
        f"{task.priority} | "
        f"{task.description}"
    )
```

```python
def create_agent_plan(user_request: str) -> AgentPlan:
    prompt = f"""
You are the Orchestrator Agent in an AI multi-agent system.

Your responsibility is to break a complex user request
into clear, ordered tasks and assign each task to the
appropriate specialized agent.

Available agents:
- research
- analysis
- retrieval
- verification
- synthesis
- automation

Rules:
1. Do not perform the tasks yourself.
2. Assign each task to the most appropriate agent.
3. Keep the plan focused on the user's objective.
4. Order tasks logically.
5. Use verification before final synthesis when factual
   claims or external evidence are involved.
6. Use automation only when an actual external action is required.

User request:
{user_request}
"""

    return generate_structured_response(
        prompt=prompt,
        response_model=AgentPlan,
    )


print("✓ Orchestrator planning function created")
```

```python
test_request = """
Build a research report about how RAG improves
the reliability of AI agents. Find relevant evidence,
analyze the benefits and limitations, verify important
claims, and produce a final report.
"""

plan = create_agent_plan(test_request)

print("✓ Orchestrator generated a plan")
print()
print("Objective:")
print(plan.objective)

print()
print("Execution Plan:")

for index, task in enumerate(plan.tasks, start=1):
    print(
        f"{index}. "
        f"[{task.agent}] "
        f"{task.description} "
        f"(priority: {task.priority})"
    )
```

```python
def validate_agent_plan(plan: AgentPlan) -> bool:
    if not plan.objective.strip():
        return False

    if not plan.tasks:
        return False

    for task in plan.tasks:
        if not task.task_id.strip():
            return False

        if not task.description.strip():
            return False

        if task.agent not in {
            "research",
            "analysis",
            "retrieval",
            "verification",
            "synthesis",
            "automation",
        }:
            return False

    return True


if validate_agent_plan(plan):
    print("✓ Agent plan passed validation")
else:
    print("✗ Agent plan failed validation")
```

```python
class OrchestratorResult(BaseModel):
    objective: str
    tasks: list[AgentTask]
    valid: bool
    task_count: int


def run_orchestrator(user_request: str) -> OrchestratorResult:
    plan = create_agent_plan(user_request)

    is_valid = validate_agent_plan(plan)

    return OrchestratorResult(
        objective=plan.objective,
        tasks=plan.tasks,
        valid=is_valid,
        task_count=len(plan.tasks),
    )


print("✓ Orchestrator result model created")
```
