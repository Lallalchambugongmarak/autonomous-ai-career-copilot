````python
# ============================================
# AI CAREER COPILOT
# PHASE 3 - LLM LAYER
# CELL 6 - STRUCTURED JSON OUTPUT
# ============================================

import json


def extract_response_text(response) -> str:
    """
    Convert Gemini/LangChain response content
    into plain text.
    """

    content = response.content

    # Normal string response
    if isinstance(content, str):
        return content.strip()

    # List of content blocks
    if isinstance(content, list):
        text_parts = []

        for block in content:
            if isinstance(block, str):
                text_parts.append(block)

            elif isinstance(block, dict):
                if "text" in block:
                    text_parts.append(str(block["text"]))

        return "\n".join(text_parts).strip()

    # Fallback
    return str(content).strip()


def run_agent_json(agent_name: str, user_input: str) -> dict:
    """
    Run a Career Copilot agent and return
    the response as a Python dictionary.
    """

    prompt = build_agent_prompt(
        agent_name=agent_name,
        user_input=user_input
    )

    structured_prompt = f"""
{prompt}

IMPORTANT:
Return ONLY valid JSON.
Do not use Markdown.
Do not use ```json or ```.

Use exactly this structure:

{{
    "agent": "{agent_name}",
    "status": "success",
    "result": {{
        "summary": "",
        "items": []
    }}
}}
"""

    response = career_llm.invoke(structured_prompt)

    # Safely extract text from Gemini response
    response_text = extract_response_text(response)

    # Remove accidental Markdown code fences
    if response_text.startswith("```"):
        response_text = response_text.replace("```json", "")
        response_text = response_text.replace("```", "")
        response_text = response_text.strip()

    try:
        return json.loads(response_text)

    except json.JSONDecodeError as e:
        return {
            "agent": agent_name,
            "status": "error",
            "result": {
                "summary": "LLM returned invalid JSON.",
                "items": []
            },
            "error": str(e),
            "raw_response": response_text
        }


# ============================================
# TEST STRUCTURED OUTPUT
# ============================================

test_json_response = run_agent_json(
    "resume_agent",
    """
    Identify the main technical skills from this profile:

    Python, LangChain, LangGraph, RAG, FastAPI,
    PostgreSQL, Docker.
    """
)

print("=" * 60)
print("STRUCTURED JSON OUTPUT TEST")
print("=" * 60)

print(json.dumps(test_json_response, indent=4))

print("=" * 60)

if test_json_response.get("status") == "success":
    print("✓ Structured JSON response received")
    print("✓ JSON successfully parsed")
else:
    print("⚠ JSON response requires review")

print("=" * 60)
````

````python
# Phase 5 — Cell 3
# Resume Analysis Tool

def resume_analysis_tool(
    resume_text: str,
    target_role: str
) -> dict[str, Any]:
    """
    Analyze resume content against a target career role.
    """

    resume_lower = resume_text.lower()

    # Core AI/Agentic AI skills to detect
    skill_keywords = [
        "python",
        "llm",
        "llms",
        "langchain",
        "langgraph",
        "rag",
        "retrieval augmented generation",
        "vector database",
        "machine learning",
        "deep learning",
        "generative ai",
        "generative artificial intelligence",
        "api",
        "fastapi",
        "docker",
        "aws",
        "azure",
        "gcp"
    ]

    detected_skills = []

    for skill in skill_keywords:
        if skill in resume_lower:
            detected_skills.append(skill)

    # Skills especially relevant to Agentic AI
    priority_skills = [
        "python",
        "llm",
        "langchain",
        "langgraph",
        "rag",
        "vector database",
        "api",
        "fastapi",
        "docker",
        "aws"
    ]

    missing_skills = [
        skill
        for skill in priority_skills
        if skill not in detected_skills
    ]

    analysis = {
        "target_role": target_role,
        "detected_skills": detected_skills,
        "missing_skills": missing_skills,
        "skill_match_count": len(detected_skills),
        "priority_gap_count": len(missing_skills)
    }

    return analysis


# Register the tool
register_career_tool(
    "resume_analysis",
    resume_analysis_tool
)


# Test resume analysis
sample_resume_text = """
AI Engineer with experience in Python, LLM APIs,
LangGraph, LangChain, RAG, FastAPI and Docker.
Built multi-agent AI applications using LLMs.
"""

phase5_resume_analysis = resume_analysis_tool(
    resume_text=sample_resume_text,
    target_role="Agentic AI Engineer"
)


print("RESUME ANALYSIS TOOL TEST")
print("=" * 60)

print("Target Role:")
print(phase5_resume_analysis["target_role"])

print("\nDetected Skills:")
print(phase5_resume_analysis["detected_skills"])

print("\nMissing Priority Skills:")
print(phase5_resume_analysis["missing_skills"])

print("\nSkill Match Count:")
print(phase5_resume_analysis["skill_match_count"])

print("\nPriority Gap Count:")
print(phase5_resume_analysis["priority_gap_count"])


# Validation
assert isinstance(phase5_resume_analysis, dict)
assert "detected_skills" in phase5_resume_analysis
assert "missing_skills" in phase5_resume_analysis
assert "resume_analysis" in career_tools

print("\nRegistered Tools:")
print(list_career_tools())

print("\n✅ Resume Analysis Tool created and registered successfully.")
````

````python
# Phase 5 — Cell 6
# Agent Tool Selection

def select_career_tool(
    user_request: str
) -> dict[str, Any]:
    """
    Select the most relevant Career Copilot tool
    based on the user's request.
    """

    request = user_request.lower()

    # Tool-selection rules
    if any(
        keyword in request
        for keyword in [
            "job",
            "jobs",
            "vacancy",
            "opening",
            "hiring",
            "career opportunities"
        ]
    ):
        selected_tool = "job_search"

    elif any(
        keyword in request
        for keyword in [
            "resume",
            "cv",
            "curriculum vitae"
        ]
    ):
        selected_tool = "resume_analysis"

    elif any(
        keyword in request
        for keyword in [
            "skill",
            "technology",
            "technologies",
            "learn",
            "framework"
        ]
    ):
        selected_tool = "skill_research"

    else:
        selected_tool = None

    return {
        "user_request": user_request,
        "selected_tool": selected_tool,
        "tool_available": (
            selected_tool in career_tools
            if selected_tool
            else False
        )
    }


# Test 1 — Job Search
phase5_tool_test_1 = select_career_tool(
    "Find Agentic AI Engineer jobs in Bangalore"
)

# Test 2 — Resume Analysis
phase5_tool_test_2 = select_career_tool(
    "Analyze my resume for an AI Engineer role"
)

# Test 3 — Skill Research
phase5_tool_test_3 = select_career_tool(
    "What technology skills should I learn for Agentic AI?"
)


print("AGENT TOOL SELECTION TEST")
print("=" * 60)

print("\nRequest 1:")
print(phase5_tool_test_1["user_request"])
print("Selected Tool:", phase5_tool_test_1["selected_tool"])

print("\nRequest 2:")
print(phase5_tool_test_2["user_request"])
print("Selected Tool:", phase5_tool_test_2["selected_tool"])

print("\nRequest 3:")
print(phase5_tool_test_3["user_request"])
print("Selected Tool:", phase5_tool_test_3["selected_tool"])


# Validation
assert phase5_tool_test_1["selected_tool"] == "job_search"
assert phase5_tool_test_2["selected_tool"] == "resume_analysis"
assert phase5_tool_test_3["selected_tool"] == "skill_research"

assert phase5_tool_test_1["tool_available"] is True
assert phase5_tool_test_2["tool_available"] is True
assert phase5_tool_test_3["tool_available"] is True

print("\n✅ Agent Tool Selection test passed.")
print("✅ The Career Copilot can identify the appropriate tool.")
````

````python
# Phase 5 — Cell 7
# Tool-Calling Agent

def career_tool_calling_agent(
    user_request: str,
    target_role: str = "Agentic AI Engineer",
    location: str = "Bangalore",
    resume_text: str = "",
    skill: str = ""
) -> dict[str, Any]:
    """
    Select and execute the appropriate Career Copilot tool.
    """

    # Step 1: Select the tool
    selection = select_career_tool(user_request)

    selected_tool = selection["selected_tool"]

    # Step 2: Handle unknown requests
    if selected_tool is None:
        return {
            "user_request": user_request,
            "selected_tool": None,
            "tool_executed": False,
            "result": {
                "message": "No suitable career tool was found."
            }
        }

    # Step 3: Retrieve the tool
    tool = get_career_tool(selected_tool)

    if tool is None:
        return {
            "user_request": user_request,
            "selected_tool": selected_tool,
            "tool_executed": False,
            "result": {
                "message": "Selected tool is not available."
            }
        }

    # Step 4: Execute the selected tool
    if selected_tool == "job_search":

        result = tool(
            target_role=target_role,
            location=location
        )

    elif selected_tool == "resume_analysis":

        result = tool(
            resume_text=resume_text,
            target_role=target_role
        )

    elif selected_tool == "skill_research":

        result = tool(
            skill=skill or target_role
        )

    else:

        result = {
            "message": "Tool execution is not configured."
        }

    return {
        "user_request": user_request,
        "selected_tool": selected_tool,
        "tool_executed": True,
        "result": result
    }


# Test the Tool-Calling Agent

phase5_tool_call_result = career_tool_calling_agent(
    user_request="Find Agentic AI Engineer jobs in Bangalore",
    target_role="Agentic AI Engineer",
    location="Bangalore"
)


print("TOOL-CALLING AGENT TEST")
print("=" * 60)

print("\nUser Request:")
print(phase5_tool_call_result["user_request"])

print("\nSelected Tool:")
print(phase5_tool_call_result["selected_tool"])

print("\nTool Executed:")
print(phase5_tool_call_result["tool_executed"])

print("\nTool Result:")
print(phase5_tool_call_result["result"])


# Validation
assert phase5_tool_call_result["selected_tool"] == "job_search"
assert phase5_tool_call_result["tool_executed"] is True
assert isinstance(
    phase5_tool_call_result["result"],
    list
)
assert len(phase5_tool_call_result["result"]) > 0

print("\n✅ Tool-Calling Agent test passed.")
print("✅ Tool selection and execution are working.")
````

````python
# Phase 7 — Cell 7
# Create the Autonomous Agent Router

print("🧭 Creating autonomous agent router...")

def phase7_route_agent(state: OrchestratorState) -> str:
    """
    Select the next specialist agent based on the current orchestration state.
    """

    pending_agents = state.get("pending_agents", [])
    completed_agents = state.get("completed_agents", [])
    cycle_count = state.get("cycle_count", 0)
    max_cycles = state.get("max_cycles", 10)

    if cycle_count >= max_cycles:
        return "END"

    if not pending_agents:
        return "END"

    for agent_name in pending_agents:
        if agent_name not in completed_agents:
            return agent_name

    return "END"


phase7_router_test_state: OrchestratorState = {
    "current_agent": None,
    "completed_agents": [],
    "pending_agents": list(phase7_agent_registry.keys()),
    "task_queue": list(phase7_agent_registry.keys()),
    "execution_history": [],
    "routing_reason": "Initial routing test.",
    "cycle_count": 0,
    "max_cycles": 10,
    "orchestration_status": "ready"
}

phase7_next_agent = phase7_route_agent(phase7_router_test_state)

assert callable(phase7_route_agent)
assert phase7_next_agent == "career_agent"

print("✅ Autonomous router created.")
print("✅ Router function is callable.")
print(f"✅ First selected agent: {phase7_next_agent}")
print("✅ Routing decision validated.")

print("\n🎯 Autonomous routing logic is operational.")
print("🎉 Phase 7 Cell 7 Complete.")
````

````python
# Phase 7 — Cell 9
# Create the Specialist Agent Execution Engine

print("⚙️ Creating specialist agent execution engine...")

def phase7_execute_agent(
    agent_name: str,
    state: MultiAgentCareerState,
    task_message: AgentMessage
) -> Dict[str, Any]:

    assert agent_name in phase7_agent_registry, \
        f"Unknown agent: {agent_name}"

    agent_results = state.get("agent_results", {}).copy()

    target_role = state.get("target_role", "")
    skills = state.get("skills", [])
    user_profile = state.get("user_profile", {})

    if agent_name == "career_agent":
        result = {
            "agent": agent_name,
            "role": "Career Strategy Specialist",
            "target_role": target_role,
            "analysis": (
                f"Career direction analysis completed for the target role: "
                f"{target_role}."
            ),
            "status": "completed"
        }

    elif agent_name == "skill_agent":
        result = {
            "agent": agent_name,
            "role": "Skill Gap Specialist",
            "current_skills": skills,
            "analysis": (
                f"Skill-gap analysis completed for the target role: "
                f"{target_role}."
            ),
            "status": "completed"
        }

    elif agent_name == "roadmap_agent":
        result = {
            "agent": agent_name,
            "role": "Career Roadmap Specialist",
            "analysis": (
                f"Career roadmap generated for the target role: "
                f"{target_role}."
            ),
            "status": "completed"
        }

    elif agent_name == "job_market_agent":
        result = {
            "agent": agent_name,
            "role": "Job Market Specialist",
            "analysis": (
                f"Job-market requirements analysis completed for: "
                f"{target_role}."
            ),
            "status": "completed"
        }

    elif agent_name == "resume_agent":
        result = {
            "agent": agent_name,
            "role": "Resume Specialist",
            "analysis": (
                f"Resume alignment analysis completed for: "
                f"{target_role}."
            ),
            "status": "completed"
        }

    elif agent_name == "critic_agent":
        result = {
            "agent": agent_name,
            "role": "Validation Specialist",
            "analysis": "Agent outputs reviewed for consistency and completeness.",
            "status": "completed"
        }

    else:
        raise ValueError(f"Unsupported agent: {agent_name}")

    agent_results[agent_name] = result

    return {
        "agent_name": agent_name,
        "result": result,
        "agent_results": agent_results,
        "status": "completed"
    }


phase7_execution_test_state: MultiAgentCareerState = {
    "user_profile": {
        "name": "Test User",
        "experience": "Entry Level"
    },
    "target_role": "AI Engineer",
    "skills": ["Python", "Machine Learning"],
    "agent_results": {}
}

phase7_execution_test = phase7_execute_agent(
    "career_agent",
    phase7_execution_test_state,
    phase7_dispatch_test
)

assert isinstance(phase7_execution_test, dict)
assert phase7_execution_test["agent_name"] == "career_agent"
assert phase7_execution_test["status"] == "completed"
assert "career_agent" in phase7_execution_test["agent_results"]

print("✅ Specialist agent execution engine created.")
print("✅ Career agent execution tested successfully.")
print("✅ Agent result stored in shared state.")

print("\n🎯 Specialist agents can now execute dispatched tasks.")
print("🎉 Phase 7 Cell 9 Complete.")
````

````python
# Phase 7 — Cell 10
# Create the Autonomous Agent Cycle

print("🔄 Creating autonomous agent execution cycle...")

def phase7_run_agent_cycle(
    agent_name: str,
    state: MultiAgentCareerState
) -> Dict[str, Any]:

    # Step 1 — Dispatch task
    task_message = phase7_dispatch_task(
        agent_name,
        state
    )

    # Step 2 — Record incoming task message
    messages = list(state.get("agent_messages", []))
    messages.append(task_message)

    # Step 3 — Execute specialist agent
    execution_result = phase7_execute_agent(
        agent_name,
        state,
        task_message
    )

    # Step 4 — Create result message
    result_message: AgentMessage = {
        "sender": agent_name,
        "receiver": "orchestrator",
        "message_type": "result",
        "task": task_message["task"],
        "data": execution_result["result"],
        "status": "completed",
        "iteration": state.get("iteration_count", 0)
    }

    messages.append(result_message)

    # Step 5 — Update shared state
    updated_state = state.copy()

    updated_state["agent_messages"] = messages
    updated_state["agent_results"] = execution_result["agent_results"]
    updated_state["iteration_count"] = (
        state.get("iteration_count", 0) + 1
    )

    return {
        "agent_name": agent_name,
        "task_message": task_message,
        "result_message": result_message,
        "state": updated_state,
        "status": "completed"
    }


phase7_cycle_test_state: MultiAgentCareerState = {
    "user_profile": {
        "name": "Test User",
        "experience": "Entry Level"
    },
    "target_role": "AI Engineer",
    "skills": ["Python", "Machine Learning"],
    "agent_messages": [],
    "agent_results": {},
    "iteration_count": 0
}

phase7_cycle_test = phase7_run_agent_cycle(
    "career_agent",
    phase7_cycle_test_state
)

assert isinstance(phase7_cycle_test, dict)
assert phase7_cycle_test["status"] == "completed"
assert phase7_cycle_test["agent_name"] == "career_agent"

assert len(phase7_cycle_test["state"]["agent_messages"]) == 2
assert "career_agent" in phase7_cycle_test["state"]["agent_results"]
assert phase7_cycle_test["state"]["iteration_count"] == 1

print("✅ Autonomous agent cycle created.")
print("✅ Task dispatch → execution → result flow verified.")
print("✅ Agent-to-orchestrator result message verified.")
print("✅ Shared state updated successfully.")

print("\n🎯 Autonomous agent execution cycle is operational.")
print("🎉 Phase 7 Cell 10 Complete.")
````

````python
# Phase 7 — Cell 11
# Create the Autonomous Orchestrator State Updater

print("🧠 Creating autonomous orchestrator state updater...")

def phase7_update_orchestrator(
    orchestrator_state: OrchestratorState,
    cycle_result: Dict[str, Any]
) -> OrchestratorState:

    agent_name = cycle_result["agent_name"]

    updated_state = orchestrator_state.copy()

    completed_agents = list(
        updated_state.get("completed_agents", [])
    )

    pending_agents = list(
        updated_state.get("pending_agents", [])
    )

    task_queue = list(
        updated_state.get("task_queue", [])
    )

    execution_history = list(
        updated_state.get("execution_history", [])
    )

    # Mark agent as completed
    if agent_name not in completed_agents:
        completed_agents.append(agent_name)

    # Remove completed agent from pending queue
    if agent_name in pending_agents:
        pending_agents.remove(agent_name)

    if agent_name in task_queue:
        task_queue.remove(agent_name)

    # Record execution history
    execution_history.append({
        "agent": agent_name,
        "status": cycle_result["status"],
        "cycle": updated_state.get("cycle_count", 0) + 1
    })

    updated_state["current_agent"] = agent_name
    updated_state["completed_agents"] = completed_agents
    updated_state["pending_agents"] = pending_agents
    updated_state["task_queue"] = task_queue
    updated_state["execution_history"] = execution_history
    updated_state["cycle_count"] = (
        updated_state.get("cycle_count", 0) + 1
    )
    updated_state["routing_reason"] = (
        f"{agent_name} completed successfully. "
        "Selecting the next pending specialist."
    )

    if not pending_agents:
        updated_state["orchestration_status"] = "completed"
    else:
        updated_state["orchestration_status"] = "running"

    return updated_state


phase7_orchestrator_test = phase7_update_orchestrator(
    phase7_orchestrator_state,
    phase7_cycle_test
)

assert isinstance(phase7_orchestrator_test, dict)
assert "career_agent" in phase7_orchestrator_test["completed_agents"]
assert "career_agent" not in phase7_orchestrator_test["pending_agents"]
assert "career_agent" not in phase7_orchestrator_test["task_queue"]

assert phase7_orchestrator_test["cycle_count"] == 1
assert len(phase7_orchestrator_test["execution_history"]) == 1
assert phase7_orchestrator_test["orchestration_status"] == "running"

print("✅ Orchestrator state updater created.")
print("✅ Completed agent removed from pending queue.")
print("✅ Execution history recorded.")
print("✅ Cycle counter updated.")
print("✅ Autonomous routing state verified.")

print("\n🎯 Orchestrator can now track agent execution automatically.")
print("🎉 Phase 7 Cell 11 Complete.")
````

````python
# Phase 7 — Cell 12
# Create the Integrated Autonomous Orchestration Cycle

print("🤖 Creating integrated autonomous orchestration cycle...")

def phase7_autonomous_orchestration_step(
    state: MultiAgentCareerState,
    orchestrator_state: OrchestratorState
) -> Dict[str, Any]:

    # Step 1 — Select next specialist
    next_agent = phase7_route_agent(orchestrator_state)

    if next_agent == "END":
        return {
            "status": "completed",
            "next_agent": "END",
            "state": state,
            "orchestrator_state": orchestrator_state
        }

    # Step 2 — Execute selected specialist
    cycle_result = phase7_run_agent_cycle(
        next_agent,
        state
    )

    # Step 3 — Update shared multi-agent state
    updated_state = cycle_result["state"]

    # Step 4 — Update orchestrator state
    updated_orchestrator_state = phase7_update_orchestrator(
        orchestrator_state,
        cycle_result
    )

    return {
        "status": "running",
        "next_agent": next_agent,
        "cycle_result": cycle_result,
        "state": updated_state,
        "orchestrator_state": updated_orchestrator_state
    }


phase7_integrated_state: MultiAgentCareerState = {
    "user_profile": {
        "name": "Test User",
        "experience": "Entry Level"
    },
    "target_role": "AI Engineer",
    "skills": ["Python", "Machine Learning"],
    "agent_messages": [],
    "agent_results": {},
    "iteration_count": 0
}

phase7_integrated_orchestrator = {
    "current_agent": None,
    "completed_agents": [],
    "pending_agents": [
        "career_agent",
        "skill_agent",
        "roadmap_agent",
        "job_market_agent",
        "resume_agent",
        "critic_agent"
    ],
    "task_queue": [
        "career_agent",
        "skill_agent",
        "roadmap_agent",
        "job_market_agent",
        "resume_agent",
        "critic_agent"
    ],
    "execution_history": [],
    "routing_reason": "Initial orchestration cycle.",
    "cycle_count": 0,
    "max_cycles": 10,
    "orchestration_status": "initialized"
}

phase7_integrated_test = phase7_autonomous_orchestration_step(
    phase7_integrated_state,
    phase7_integrated_orchestrator
)

assert isinstance(phase7_integrated_test, dict)
assert phase7_integrated_test["status"] == "running"
assert phase7_integrated_test["next_agent"] == "career_agent"

assert "career_agent" in (
    phase7_integrated_test["state"]["agent_results"]
)

assert "career_agent" in (
    phase7_integrated_test["orchestrator_state"]["completed_agents"]
)

assert "career_agent" not in (
    phase7_integrated_test["orchestrator_state"]["pending_agents"]
)

assert (
    phase7_integrated_test["orchestrator_state"]["cycle_count"] == 1
)

print("✅ Integrated orchestration cycle created.")
print("✅ Router selected the next specialist.")
print("✅ Dispatcher delivered the task.")
print("✅ Specialist agent executed successfully.")
print("✅ Orchestrator state updated automatically.")
print("✅ Shared multi-agent state updated successfully.")

print("\n🎯 Autonomous orchestration pipeline is operational.")
print("🎉 Phase 7 Cell 12 Complete.")
````

````python
# Phase 7 — Cell 14
# Create the Multi-Agent Critic and Validation Loop

print("🔍 Creating critic and validation loop...")

def phase7_validate_agent_results(
    state: MultiAgentCareerState
) -> Dict[str, Any]:

    agent_results = state.get("agent_results", {})

    required_agents = [
        "career_agent",
        "skill_agent",
        "roadmap_agent",
        "job_market_agent",
        "resume_agent",
        "critic_agent"
    ]

    missing_agents = [
        agent for agent in required_agents
        if agent not in agent_results
    ]

    validation_checks = {
        "all_required_agents_present": len(missing_agents) == 0,
        "results_are_dicts": all(
            isinstance(result, dict)
            for result in agent_results.values()
        ),
        "all_results_completed": all(
            result.get("status") == "completed"
            for result in agent_results.values()
        )
    }

    validation_passed = (
        len(missing_agents) == 0
        and all(validation_checks.values())
    )

    validation_result = {
        "validation_passed": validation_passed,
        "checks": validation_checks,
        "missing_agents": missing_agents,
        "reviewed_agents": list(agent_results.keys()),
        "message": (
            "All specialist outputs passed validation."
            if validation_passed
            else "Validation identified incomplete specialist outputs."
        )
    }

    return validation_result


phase7_validation_test = phase7_validate_agent_results(
    phase7_full_state
)

assert isinstance(phase7_validation_test, dict)
assert "validation_passed" in phase7_validation_test
assert phase7_validation_test["validation_passed"] is True
assert phase7_validation_test["missing_agents"] == []

print("✅ Critic validation engine created.")
print("✅ Required specialist outputs verified.")
print("✅ Result structure validation passed.")
print("✅ Completion status validation passed.")
print("✅ All specialist outputs passed validation.")

print("\n🎯 Multi-agent critic/validation loop is operational.")
print("🎉 Phase 7 Cell 14 Complete.")
````

````python
# Phase 8 — Cell 9
# Create the Real Agent Execution Wrapper

print("⚙️ Building real-agent execution wrapper...")

def phase8_execute_real_agent(agent_name, state):
    """
    Execute the connected reasoning node for a Phase 8 specialist agent.
    """

    assert agent_name in phase8_agent_execution_registry, (
        f"Unknown Phase 8 agent: {agent_name}"
    )

    agent_config = phase8_agent_execution_registry[agent_name]
    execution_node = agent_config.get("execution_node")

    if execution_node is None:
        return {
            "agent": agent_name,
            "status": "pending",
            "result": None,
            "source": agent_config.get("source")
        }

    try:
        result = execution_node(state)

        return {
            "agent": agent_name,
            "status": "completed",
            "result": result,
            "source": agent_config.get("source")
        }

    except Exception as e:
        return {
            "agent": agent_name,
            "status": "error",
            "result": None,
            "source": agent_config.get("source"),
            "error": str(e)
        }


assert callable(phase8_execute_real_agent)

print("✅ Real-agent execution wrapper created.")
print("✅ Connected execution nodes can now be invoked through one interface.")
print("🎯 Phase 8 Cell 9 Complete.")
````

````python
# Phase 8 — Cell 13
# Build Shared-State Integration for Real Agent Results

print("🔗 Building shared-state integration layer...")

def phase8_store_agent_result(state, agent_name, execution_result):
    """
    Store a real specialist-agent result in the shared multi-agent state.
    """

    updated_state = dict(state)

    agent_results = dict(
        updated_state.get("agent_results", {})
    )

    agent_messages = list(
        updated_state.get("agent_messages", [])
    )

    agent_results[agent_name] = execution_result

    agent_messages.append({
        "sender": agent_name,
        "receiver": "orchestrator",
        "message_type": "result",
        "task": f"Execute {agent_name}",
        "data": execution_result.get("result"),
        "status": execution_result.get("status"),
        "iteration": updated_state.get("iteration_count", 0)
    })

    updated_state["agent_results"] = agent_results
    updated_state["agent_messages"] = agent_messages

    return updated_state


assert callable(phase8_store_agent_result)

phase8_shared_state_test = {
    "user_profile": {},
    "target_role": "AI Engineer",
    "skills": [],
    "agent_messages": [],
    "agent_results": {},
    "iteration_count": 0
}

phase8_shared_state_test = phase8_store_agent_result(
    phase8_shared_state_test,
    "career_agent",
    phase8_all_execution_results["career_agent"]
)

assert "career_agent" in phase8_shared_state_test["agent_results"]
assert len(phase8_shared_state_test["agent_messages"]) == 1

print("✅ Agent result stored in shared state.")
print("✅ Agent message created for orchestrator.")
print("✅ Shared state remains compatible with Phase 7 architecture.")

print("\n🎯 Shared-state integration verified.")
print("🎉 Phase 8 Cell 13 Complete.")
````

````python
# Phase 8 — Cell 15
# Integrate Critic Validation into Shared Multi-Agent State

print("🧪 Integrating Critic validation loop...")

def phase8_run_critic_validation(state):
    """
    Run the Critic Agent against the current shared multi-agent state.
    """

    critic_result = phase8_execute_real_agent(
        "critic_agent",
        state
    )

    updated_state = phase8_store_agent_result(
        state,
        "critic_agent",
        critic_result
    )

    updated_state["validation_result"] = critic_result

    if critic_result["status"] == "completed":
        updated_state["validation_passed"] = True
    else:
        updated_state["validation_passed"] = False

    return updated_state


assert callable(phase8_run_critic_validation)

phase8_validation_state = phase8_multi_agent_state.copy()

phase8_validation_state = phase8_run_critic_validation(
    phase8_validation_state
)

assert "validation_result" in phase8_validation_state
assert "validation_passed" in phase8_validation_state
assert phase8_validation_state["validation_result"]["status"] == "completed"
assert phase8_validation_state["validation_passed"] is True

print("✅ Critic Agent executed successfully.")
print("✅ Validation result stored in shared state.")
print("✅ Validation status recorded.")
print(
    f"✅ validation_passed = "
    f"{phase8_validation_state['validation_passed']}"
)

print("\n🎯 Critic validation loop verified.")
print("🎉 Phase 8 Cell 15 Complete.")
````

````python
# Phase 8 — Cell 16
# Build Conditional Critic Routing

print("🧭 Building autonomous validation-based routing...")

def phase8_route_after_validation(state):
    """
    Decide the next action based on Critic validation.
    """

    validation_passed = state.get(
        "validation_passed",
        False
    )

    iteration_count = state.get(
        "iteration_count",
        0
    )

    max_iterations = 3

    if validation_passed:
        return "FINALIZE"

    if iteration_count >= max_iterations:
        return "FINALIZE_WITH_WARNING"

    return "REVISE"


phase8_validation_routes = {
    "FINALIZE": "Validation passed — finalize response.",
    "REVISE": "Validation failed — request specialist revision.",
    "FINALIZE_WITH_WARNING": (
        "Maximum validation iterations reached — "
        "finalize with warning."
    )
}

assert callable(phase8_route_after_validation)

phase8_route_test_passed = phase8_route_after_validation(
    phase8_validation_state
)

assert phase8_route_test_passed == "FINALIZE"

print(f"✅ Validation route selected: {phase8_route_test_passed}")
print("✅ Conditional routing function verified.")
print("✅ Critic result now controls the next workflow action.")

print("\n🎯 Autonomous validation routing verified.")
print("🎉 Phase 8 Cell 16 Complete.")
````

````python
# Phase 8 — Cell 17
# Build Autonomous Revision / Feedback Loop

print("🔄 Building autonomous revision feedback loop...")

def phase8_prepare_revision(state):
    """
    Prepare shared state for another specialist-agent revision cycle.
    """

    updated_state = dict(state)

    current_iteration = updated_state.get(
        "iteration_count",
        0
    )

    updated_state["iteration_count"] = current_iteration + 1

    updated_state["validation_passed"] = False

    agent_messages = list(
        updated_state.get("agent_messages", [])
    )

    agent_messages.append({
        "sender": "critic_agent",
        "receiver": "orchestrator",
        "message_type": "feedback",
        "task": "Request specialist revision",
        "data": updated_state.get("validation_result"),
        "status": "revision_requested",
        "iteration": updated_state["iteration_count"]
    })

    updated_state["agent_messages"] = agent_messages

    return updated_state


assert callable(phase8_prepare_revision)

phase8_revision_test_state = {
    **phase8_validation_state,
    "validation_passed": False,
    "iteration_count": 0
}

phase8_revision_test_state = phase8_prepare_revision(
    phase8_revision_test_state
)

assert phase8_revision_test_state["iteration_count"] == 1
assert phase8_revision_test_state["validation_passed"] is False

revision_messages = [
    message
    for message in phase8_revision_test_state["agent_messages"]
    if message.get("message_type") == "feedback"
]

assert len(revision_messages) >= 1

print("✅ Revision cycle increments iteration count.")
print("✅ Validation state reset for revision.")
print("✅ Critic feedback message added to shared state.")
print("✅ Specialist revision can now be requested.")

print("\n🎯 Autonomous feedback loop verified.")
print("🎉 Phase 8 Cell 17 Complete.")
````

````python
# Phase 8 — Cell 18
# Build Autonomous Orchestrator Decision Step

print("🤖 Building autonomous orchestrator decision layer...")

def phase8_orchestrator_decision(state):
    """
    Decide whether to finalize or initiate another revision cycle.
    """

    route = phase8_route_after_validation(state)

    if route == "FINALIZE":
        return {
            "action": "finalize",
            "state": state
        }

    if route == "REVISE":
        revised_state = phase8_prepare_revision(state)

        return {
            "action": "revise",
            "state": revised_state
        }

    return {
        "action": "finalize_with_warning",
        "state": state
    }


assert callable(phase8_orchestrator_decision)

phase8_decision_test = phase8_orchestrator_decision(
    phase8_validation_state
)

assert phase8_decision_test["action"] == "finalize"
assert phase8_decision_test["state"]["validation_passed"] is True

print(
    f"✅ Orchestrator action: "
    f"{phase8_decision_test['action']}"
)

print("✅ Validation result controls orchestrator behavior.")
print("✅ Finalization path verified.")
print("✅ Revision path is available when validation fails.")

print("\n🎯 Autonomous orchestrator decision layer verified.")
print("🎉 Phase 8 Cell 18 Complete.")
````

````python
# Phase 9 — Cell 9
# Create shared specialist state manager

print("🧠 Building Phase 9 shared specialist state...")

def phase9_store_specialist_result(
    state,
    agent_name,
    execution_result
):
    """
    Store a specialist result in the shared Phase 9 state.
    """

    updated_state = dict(state)

    # Preserve existing results
    agent_results = dict(
        updated_state.get("agent_results", {})
    )

    # Preserve existing messages
    agent_messages = list(
        updated_state.get("agent_messages", [])
    )

    # Store specialist result
    agent_results[agent_name] = execution_result

    # Create inter-agent message
    agent_messages.append({
        "sender": agent_name,
        "receiver": "orchestrator",
        "message_type": "result",
        "task": f"Phase 9 specialist execution: {agent_name}",
        "data": execution_result.get("result"),
        "status": execution_result.get("status"),
        "iteration": updated_state.get(
            "iteration_count",
            0
        )
    })

    updated_state["agent_results"] = agent_results
    updated_state["agent_messages"] = agent_messages

    return updated_state


assert callable(
    phase9_store_specialist_result
)

print("✅ Shared specialist state manager created.")
print("✅ Agent results can now be stored centrally.")
print("✅ Inter-agent messages can now be recorded.")
print("🎯 Phase 9 shared-state mechanism ready.")
print("🎉 Phase 9 Cell 9 Complete.")
````
